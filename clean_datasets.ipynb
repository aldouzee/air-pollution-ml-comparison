```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# I. Clean WB WDI dataset

```python
df = pd.read_csv("WB_Extracted_WDI_Data.csv")
df
```

```python
df.columns
```

```python
# change ".." to nan
df.replace("..", np.nan, inplace=True)
df.isnull().sum()
```

```python
df.info()
```

```python
# drop columns with empty rows
df_clean = df.dropna(subset=['Series Code'])
```

```python
# reshape from wide to long format
year_cols = ['2021 [YR2021]', '2022 [YR2022]', '2023 [YR2023]', '2024 [YR2024]', '2025 [YR2025]']
df_clean = df_clean.melt(
    id_vars=['Country Name', 'Country Code', 'Series Name', 'Series Code'],
    value_vars=year_cols,
    var_name='Year',
    value_name='Value'
)
```

```python
# clean year column
df_clean['Year'] = df_clean['Year'].str.split(' ').str[0].astype(int)
df_clean['Value'] = pd.to_numeric(df_clean['Value'], errors='coerce')
```

```python
# round off values in value column
df_clean['Value'] = df_clean['Value'].round(2)
df_clean.head()
```

```python
# save current dataset
df_clean.to_csv("WB_WDI_Cleaned.csv", index=False)
```
